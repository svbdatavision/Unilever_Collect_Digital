```python
import pdfplumber
import pandas as pd
import re
import numpy as np


def parse_remittance_universal(
    pdf_path,
    columnas_keywords,
    voucher_tokens=None,        # opcional: tokens para separar registros tipo Cenco
    min_header_similarity=0.55  # porcentaje mínimo de matching para considerar un encabezado válido
):
    """
    PARSER UNIVERSAL PARA REMITTANCE PDF

    Parámetros:
    -----------
    pdf_path : str
        Ruta al PDF.
    columnas_keywords : dict
        {"NOMBRE_COLUMNA" : ["keyword1", "keyword2", ...]}
        Las keywords se comparan contra las palabras del encabezado.
    voucher_tokens : list[str] (opcional)
        Tokens que indican comienzo de registro cuando el PDF no contiene tabla alineada (Cenco).
    min_header_similarity : float
        Porcentaje mínimo de coincidencia entre encabezado detectado y keywords esperadas.

    Retorna:
    --------
    pandas.DataFrame con las columnas definidas en `columnas_keywords`
    """

    # ==========================================================
    # 1) Intentamos detectar tabla por encabezados (modo Euro)
    # ==========================================================
    with pdfplumber.open(pdf_path) as pdf:
        for page in pdf.pages:
            words = page.extract_words()
            if not words:
                continue

            # 1A. identificar línea candidata a encabezado
            tops = [w["top"] for w in words]
            common_top = np.median(tops)

            header_line = [w for w in words if abs(w["top"] - common_top) < 4]

            if len(header_line) == 0:
                continue

            # 1B. detectar palabras del encabezado
            header_texts = [w["text"].lower() for w in header_line]

            # 1C. evaluar match con keywords
            matches = {}
            for col, keys in columnas_keywords.items():
                found = any(any(k.lower() in ht for k in keys) for ht in header_texts)
                matches[col] = found

            match_score = sum(matches.values()) / len(matches)

            if match_score >= min_header_similarity:
                # ===================================================
                # MODO TABLA DETECTADO
                # ===================================================

                # Ordenar encabezado por posición X
                header_sorted = sorted(header_line, key=lambda w: w["x0"])

                # Build columns x-centers
                col_positions = {col: None for col in columnas_keywords}

                # asignar cada columna al header más parecido
                for col, keys in columnas_keywords.items():
                    best = None
                    best_score = 0
                    for hw in header_sorted:
                        score = sum(1 for k in keys if k.lower() in hw["text"].lower())
                        if score > best_score:
                            best_score = score
                            best = hw["x0"]
                    col_positions[col] = best

                # Ordenar columnas por X
                cols_sorted = sorted(col_positions.items(), key=lambda x: x[1] if x[1] is not None else 1e9)
                colnames = [c[0] for c in cols_sorted]
                xs = [c[1] for c in cols_sorted]

                # Crear boundaries
                bounds = [0] + [(a + b) / 2 for a, b in zip(xs, xs[1:])] + [page.width + 5]

                # ===================================================
                # Extraer filas debajo del encabezado
                # ===================================================
                rows = []
                data_words = [w for w in words if w["top"] > common_top + 5]
                data_words = sorted(data_words, key=lambda w: (w["top"], w["x0"]))

                current = []
                last_top = None
                row_tol = 6

                for w in data_words:
                    if last_top is None or abs(w["top"] - last_top) <= row_tol:
                        current.append(w)
                        last_top = w["top"]
                    else:
                        rows.append(current)
                        current = [w]
                        last_top = w["top"]
                if current:
                    rows.append(current)

                # ===================================================
                # Armar dataframe
                # ===================================================
                parsed_rows = []
                for row in rows:
                    row_dict = {col: "" for col in colnames}
                    for w in row:
                        x = w["x0"]
                        idx = None
                        for i in range(len(bounds) - 1):
                            if bounds[i] <= x < bounds[i + 1]:
                                idx = i
                                break
                        if idx is None:
                            idx = len(cols_sorted) - 1

                        row_dict[colnames[idx]] += " " + w["text"]

                    parsed_rows.append({c: v.strip() for c, v in row_dict.items()})

                return pd.DataFrame(parsed_rows)

    # ==========================================================
    # 2) Si no se detectó tabla → MODO TEXTO (caso Cenco)
    # ==========================================================
    if voucher_tokens is None:
        voucher_tokens = []

    reader = PdfReader(pdf_path)
    pages_text = [p.extract_text() or "" for p in reader.pages]

    registros = []
    for text in pages_text:
        lines = [ln.strip() for ln in text.splitlines() if ln.strip() != ""]
        buffer = []

        for ln in lines:
            first = ln.split()[0] if ln.split() else ""
            if first in voucher_tokens:
                if buffer:
                    registros.append(" ".join(buffer))
                buffer = [ln]
            else:
                if buffer:
                    buffer.append(ln)
                else:
                    buffer = [ln]

        if buffer:
            registros.append(" ".join(buffer))

    df_text = pd.DataFrame({"raw": registros})

    # Extraer según columnas_keywords con regex muy general
    for col in columnas_keywords:
        df_text[col] = ""

    return df_text[columnas_keywords.keys()]
```

```python
columnas_cenco = {
    "VOUCHER": ["voucher","fs","dav","dpc","dat","dca"],
    "DESCRIPCION": ["descripcion","factura","detalle"],
    "DOCUMENTO": ["documento","pmp","pp","factura"],
    "TIENDA": ["tienda","adm","co","c.o"],
    "F. REGISTRO": ["fecha","registro"],
    "VALOR FAC.": ["valor","factura","fac"],
    "IVA FAC.": ["iva"],
    "RET. FUENTE": ["ret fuente","retencion"],
    "RET. IVA": ["ret iva"],
    "RET. ICA": ["ret ica","ica"],
    "OTROS IMP.": ["otros"],
    "VALOR PAG": ["valor pag","pago"],
    "DOC.SOPORTE": ["soporte","doc"],
}
```

```python
df_cenco = parse_remittance_universal(
    "Remittance_Cenco.pdf",
    columnas_keywords=columnas_cenco,
    voucher_tokens=["DAT","CH","DAV","DCA","DCC","DCF","DEV","DND","DPC","FPM","FS","LTG","RPL","VOUCHER"]
)

df_cenco.head(100)
```
