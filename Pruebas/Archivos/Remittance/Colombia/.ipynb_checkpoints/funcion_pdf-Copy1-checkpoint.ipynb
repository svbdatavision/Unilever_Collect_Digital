```python
import difflib
import pandas as pd
import numpy as np

# Intentamos importar herramientas opcionales (si no están, seguimos con pdfplumber/rebuild)
try:
    import camelot
except Exception:
    camelot = None

import pdfplumber


# ---------------------------
# Helpers: scoring / normalizar
# ---------------------------
def _best_header_match_for_table(df, expected_headers, max_rows_to_check=6, min_token_ratio=0.45):
    """
    Para una tabla (DataFrame), busca entre sus primeras filas (hasta max_rows_to_check)
    la fila que mejor funciona como encabezado respecto a expected_headers.
    Devuelve: (best_score_fraction, best_row_index, matches_count, total_expected)
    """
    if df is None or df.shape[0] == 0:
        return 0.0, None, 0, len(expected_headers)

    total_expected = len(expected_headers)
    best_row = None
    best_score = 0.0
    best_matches = 0

    # convertir todo a str
    df_str = df.fillna("").astype(str)

    rows_to_check = min(max_rows_to_check, df_str.shape[0])
    for r in range(rows_to_check):
        row_vals = [c.strip().lower() for c in df_str.iloc[r].tolist()]
        matches = 0
        for expected in expected_headers:
            expected_l = expected.strip().lower()
            # buscar mejor columna que parezca el expected (fuzzy)
            best_ratio = 0.0
            for val in row_vals:
                if val == "":
                    continue
                ratio = difflib.SequenceMatcher(None, expected_l, val).ratio()
                if ratio > best_ratio:
                    best_ratio = ratio
            if best_ratio >= min_token_ratio:
                matches += 1
        score = matches / max(1, total_expected)
        if score > best_score:
            best_score = score
            best_row = r
            best_matches = matches

    return best_score, best_row, best_matches, total_expected


def _map_and_normalize(df, expected_headers, header_row_idx):
    """
    Toma df y una fila header_row_idx que corresponde a los encabezados del pdf.
    Mapea columnas a expected_headers por fuzzy matching y retorna DataFrame normalizado
    con columnas en el orden de expected_headers y filas data debajo del header_row_idx.
    """
    if df is None or df.shape[0] == 0:
        return pd.DataFrame(columns=expected_headers)

    df_str = df.fillna("").astype(str).reset_index(drop=True)

    # Si header_row_idx es None -> asumimos la primera fila
    if header_row_idx is None:
        header_row_idx = 0
    header_vals = df_str.iloc[header_row_idx].tolist()

    mapping = {}  # expected -> column index in df
    used_cols = set()
    for expected in expected_headers:
        best_idx = None
        best_ratio = -1.0
        for idx, hv in enumerate(header_vals):
            if idx in used_cols:
                continue
            ratio = difflib.SequenceMatcher(None, str(expected).strip().lower(), str(hv).strip().lower()).ratio()
            if ratio > best_ratio:
                best_ratio = ratio
                best_idx = idx
        # si el mejor_ratio es muy bajo igual lo asignamos (evita None), pero puede quedar vacío
        mapping[expected] = best_idx
        if best_idx is not None:
            used_cols.add(best_idx)

    # construir dataframe resultado
    rows = []
    for i in range(header_row_idx + 1, df_str.shape[0]):
        row = {}
        for expected in expected_headers:
            idx = mapping.get(expected)
            if idx is None or idx >= df_str.shape[1]:
                row[expected] = ""
            else:
                row[expected] = df_str.iat[i, idx]
        rows.append(row)

    result = pd.DataFrame(rows, columns=expected_headers)
    # limpieza básica
    result = result.applymap(lambda x: x.strip() if isinstance(x, str) else x)
    # eliminar filas vacías
    result = result.loc[~(result == "").all(axis=1)].reset_index(drop=True)
    return result


# ---------------------------
# Extracción: camelot, pdfplumber, positional rebuild
# ---------------------------
def _extract_with_camelot(pdf_path, flavor="stream"):
    if camelot is None:
        return []
    try:
        tables = camelot.read_pdf(pdf_path, pages="all", flavor=flavor)
        dfs = []
        for t in tables:
            try:
                dfs.append(pd.DataFrame(t.df))
            except Exception:
                continue
        return dfs
    except Exception:
        return []


def _extract_with_pdfplumber_native(pdf_path):
    dfs = []
    try:
        with pdfplumber.open(pdf_path) as pdf:
            for page in pdf.pages:
                try:
                    for table in page.extract_tables():
                        dfs.append(pd.DataFrame(table))
                except Exception:
                    continue
    except Exception:
        pass
    return dfs


def _rebuild_by_positions(pdf_path):
    """
    Último recurso: reconstruye filas por posición (top/x0) usando pdfplumber.extract_words()
    Devuelve una lista con un único DataFrame (rows x words).
    """
    rows = []
    try:
        with pdfplumber.open(pdf_path) as pdf:
            for page in pdf.pages:
                words = page.extract_words()
                if not words:
                    continue
                # ordenar por top luego x0
                words_sorted = sorted(words, key=lambda w: (round(w["top"]), w["x0"]))
                current = []
                last_top = None
                tol = 6
                for w in words_sorted:
                    if last_top is None or abs(round(w["top"]) - last_top) <= tol:
                        current.append(w["text"])
                        last_top = round(w["top"])
                    else:
                        rows.append(current)
                        current = [w["text"]]
                        last_top = round(w["top"])
                if current:
                    rows.append(current)
    except Exception:
        pass

    if not rows:
        return []
    # unify into a DataFrame (ragged -> fill)
    maxc = max(len(r) for r in rows)
    norm = [r + [""] * (maxc - len(r)) for r in rows]
    return [pd.DataFrame(norm)]


# ---------------------------
# Scoring candidate tables
# ---------------------------
def _score_candidate_table(df, expected_headers, min_token_ratio=0.45):
    """
    Para cada tabla (df), devuelve (score, header_row_idx, matches_count, total_expected)
    score = fraction matched headers
    """
    score, hdr_row, matches, total = _best_header_match_for_table(df, expected_headers, max_rows_to_check=6, min_token_ratio=min_token_ratio)
    return score, hdr_row, matches, total


# ---------------------------
# Función principal
# ---------------------------
def detect_table_universal(
    pdf_path,
    expected_headers,
    min_similarity=0.45,      # umbral mínimo para aceptar tabla ganadora
    min_token_ratio=0.45,     # ratio fuzzy para considerar una columna coincidente en la fila header
    top_n=1,                  # devuelve top_n tablas mejores (por defecto 1)
    voucher_tokens=None,      # opcional: lista para filtrar por primer token en la primera columna (ej Cenco)
    require_non_empty_cols=1  # mínimo de columnas no vacías para aceptar fila (post-filtrado)
):
    """
    Motor universal:
      - genera candidates con camelot (lattice+stream), pdfplumber native y positional rebuild
      - puntúa cada candidate buscando la mejor fila header
      - normaliza y devuelve la(s) mejores tablas (DataFrame(s))

    Retorna:
      - si top_n == 1: DataFrame
      - si top_n > 1: list[DataFrame] ordenadas por score desc
    """

    expected_headers = [h.strip() for h in expected_headers]
    candidates = []

    # 1) Camelot lattice / stream
    if camelot is not None:
        candidates += [(df, "camelot_lattice") for df in _extract_with_camelot(pdf_path, flavor="lattice")]
        candidates += [(df, "camelot_stream") for df in _extract_with_camelot(pdf_path, flavor="stream")]

    # 2) pdfplumber native tables
    candidates += [(df, "plumber_native") for df in _extract_with_pdfplumber_native(pdf_path)]

    # 3) positional rebuild (last resort)
    candidates += [(df, "pos_rebuild") for df in _rebuild_by_positions(pdf_path)]

    if not candidates:
        raise ValueError("No se detectaron tablas con los extractores disponibles.")

    scored = []
    for df, source in candidates:
        try:
            score, hdr_row, matches, total = _score_candidate_table(df, expected_headers, min_token_ratio)
            scored.append({"df": df, "source": source, "score": score, "hdr_row": hdr_row, "matches": matches, "total": total})
        except Exception:
            continue

    if not scored:
        raise ValueError("No se pudo puntuar ninguna tabla candidata.")

    # ordenar por score desc
    scored_sorted = sorted(scored, key=lambda x: x["score"], reverse=True)

    # filtrar por score mínimo
    filtered = [s for s in scored_sorted if s["score"] >= min_similarity]

    if not filtered:
        # si ninguna pasa el umbral, devolvemos el mejor (si el mejor es muy malo avisamos)
        best = scored_sorted[0]
        if best["score"] < min_similarity:
            raise ValueError(f"Ninguna tabla alcanza la similitud mínima. Mejor score = {best['score']:.2f} (matches {best['matches']}/{best['total']}) from {best['source']}")
        filtered = [best]

    results = []
    for s in filtered[:top_n]:
        norm = _map_and_normalize(s["df"], expected_headers, s["hdr_row"])
        # post-filter: filas con al menos require_non_empty_cols no vacías
        if require_non_empty_cols > 0:
            mask = norm.apply(lambda row: row.astype(str).str.strip().replace("", pd.NA).count(), axis=1) >= require_non_empty_cols
            norm = norm[mask].reset_index(drop=True)
        # voucher filter (opcional)
        if voucher_tokens:
            first_col = norm.columns[0]
            first_tokens = norm[first_col].astype(str).str.strip().str.split().str[0].fillna("")
            tokens_upper = [t.upper() for t in voucher_tokens]
            norm = norm[first_tokens.str.upper().isin(tokens_upper)].reset_index(drop=True)

        results.append({"df": norm, "score": s["score"], "source": s["source"], "matches": s["matches"], "hdr_row": s["hdr_row"]})

    if not results:
        raise ValueError("Después de normalizar no quedaron filas válidas.")

    if top_n == 1:
        return results[0]["df"]
    else:
        return results
```

```python
import re
import camelot
import pandas as pd
import pdfplumber
from PyPDF2 import PdfReader


# ==========================================================
# =========== PARSEADOR 100% GENERAL Y EXTENSIBLE ==========
# ==========================================================
def parse_remittance(
    pdf_path,
    columnas_finales,
    metodo="texto",
    voucher_tokens=None,
    regex_documento=None,
    regex_fecha=None,
    regex_numeros=r"-?\d[\d\.,]*",
    camelot_flavor="stream",
    camelot_row_tol=15,
    camelot_edge_tol=300,
    extractor_custom=None, 
):
    """
    PARSEADOR GENERAL PARA CUALQUIER CLIENTE.

    Parámetros claves:
    -------------------
    columnas_finales : lista de nombres de columnas deseadas
    metodo : "texto" | "plumber" | "camelot"
    voucher_tokens : lista opcional, para cortar registros (solo Cenco)
    regex_documento : patrón opcional para capturar documento
    regex_fecha : patrón opcional para fecha
    regex_numeros : patrón opcional para valores numéricos
    extractor_custom : función alternativa (usado para Euro)

    Funcionamiento:
    ---------------
    Si se usa extractor_custom → se ejecuta ese parser.
    Si no → se usa metodo (texto / plumber / camelot).
    """

    # ==========================================================
    # ============= PRIORIDAD MÁXIMA: CUSTOM EXTRACTOR =========
    # ==========================================================
    if extractor_custom is not None:
        df = extractor_custom(pdf_path)
        return normalizar(df, columnas_finales)

    # ==========================================================
    # ============= MÉTODO 1: CAMELOT ==========================
    # ==========================================================
    if metodo == "camelot":
        try:
            tables = camelot.read_pdf(
                pdf_path,
                pages="all",
                flavor=camelot_flavor,
                row_tol=camelot_row_tol,
                edge_tol=camelot_edge_tol,
                strip_text="\n",
                split_text=True
            )
            df = pd.concat([t.df for t in tables], ignore_index=True)
        except:
            df = pd.DataFrame()

        return normalizar(df, columnas_finales)

    # ==========================================================
    # ============= MÉTODO 2: PLUMBER ==========================
    # ==========================================================
    if metodo == "plumber":
        rows = []
        with pdfplumber.open(pdf_path) as pdf:
            for page in pdf.pages:
                table = page.extract_table()
                if table:
                    rows.extend(table)
        df = pd.DataFrame(rows)
        return normalizar(df, columnas_finales)

    # ==========================================================
    # ============= MÉTODO 3: TEXTO ============================
    # ==========================================================
    if metodo == "texto":

        if voucher_tokens is None:
            voucher_tokens = []

        reader = PdfReader(pdf_path)
        pages_text = [p.extract_text() or "" for p in reader.pages]
        registros = []

        # ---- Cortado por tokens ----
        for text in pages_text:
            lines = [ln.strip() for ln in text.splitlines() if ln.strip()]
            buffer = []
            for ln in lines:
                fw = ln.split()[0] if ln.split() else ""
                if fw in voucher_tokens:
                    if buffer:
                        registros.append(" ".join(buffer))
                    buffer = [ln]
                else:
                    if buffer:
                        buffer.append(ln)
                    else:
                        buffer = [ln]
            if buffer:
                registros.append(" ".join(buffer))

        parsed = []

        for rec in registros:
            parts = rec.split()

            voucher = parts[0] if parts and parts[0] in voucher_tokens else ""

            documento = ""
            if regex_documento:
                m = re.search(regex_documento, rec)
                documento = m.group(0) if m else ""

            fecha = ""
            if regex_fecha:
                m = re.search(regex_fecha, rec)
                fecha = m.group(0) if m else ""

            nums = re.findall(regex_numeros, rec)
            nums_clean = []
            for v in nums:
                v2 = v.replace(".", "").replace(",", "")
                if re.search(r"\d", v2):
                    nums_clean.append(v)

            valor_pag = nums_clean[-2] if len(nums_clean) >= 2 else ""
            doc_soporte = nums_clean[-1] if len(nums_clean) >= 1 else ""
            descripcion = " ".join(parts[1:4]).upper() if voucher else ""

            parsed.append([
                voucher,
                descripcion,
                documento,
                "",
                fecha,
                "",
                "",
                "",
                "",
                "",
                "",
                "",
                valor_pag,
                doc_soporte
            ])

        df = pd.DataFrame(parsed)
        return normalizar(df, columnas_finales)

    raise ValueError("Método no reconocido.")


# ==========================================================
# ================ NORMALIZADOR GENERAL =====================
# ==========================================================
def normalizar(df, columnas_finales):

    if df.empty:
        return pd.DataFrame(columns=columnas_finales)

    # Ajuste de columnas
    if len(df.columns) > len(columnas_finales):
        df = df.iloc[:, :len(columnas_finales)]

    elif len(df.columns) < len(columnas_finales):
        for i in range(len(columnas_finales) - len(df.columns)):
            df[f"extra_{i}"] = ""

    df.columns = columnas_finales

    # Limpieza final
    df = df.applymap(lambda x: str(x).strip() if pd.notnull(x) else x)

    return df
```

```python
import pdfplumber
import pandas as pd
import difflib
import re


def parse_pdf_table(
    pdf_path,
    headers,
    synonyms=None,
    fuzzy_threshold=0.35,
    row_tolerance=7,
    merge_multiline=False,
    multiline_key=None
):
    """
    EXTRACTION ENGINE UNIVERSAL PARA TABLAS EN PDF
    ----------------------------------------------
    Parámetros:
        pdf_path: ruta del PDF
        headers: lista de nombres exactos de columnas
        synonyms: {header: [palabras similares]} OPCIONAL
        fuzzy_threshold: mínimo ratio para considerar un header detectado
        row_tolerance: tolerancia vertical para agrupar filas
        merge_multiline: True para unir filas multilínea (Euro)
        multiline_key: columna clave para detectar continuidad

    Funciona con:
        - Cenco
        - Euro
        - D1
        - Nuevos clientes (solo pasás headers)
    """

    synonyms = synonyms or {}
    detected = {}
    collected_rows = []

    with pdfplumber.open(pdf_path) as pdf:

        for page in pdf.pages:
            words = page.extract_words()

            # -----------------------------------------
            # 1) DETECTAR HEADERS (con fuzzy matching)
            # -----------------------------------------
            for h in headers:
                best = None
                best_ratio = 0

                candidates = [h] + synonyms.get(h, [])

                for w in words:
                    clean = re.sub(r"[^A-Za-z0-9 ]", "", w["text"]).strip()
                    if not clean:
                        continue

                    for cand in candidates:
                        ratio = difflib.SequenceMatcher(None, cand.lower(), clean.lower()).ratio()
                        if ratio > best_ratio:
                            best_ratio = ratio
                            best = w

                if best and best_ratio >= fuzzy_threshold:
                    detected[h] = (best["x0"] + best["x1"]) / 2

            if len(detected) < 2:
                continue

            # Ordenar columnas por posición
            ordered = sorted(detected.items(), key=lambda x: x[1])
            colnames = [x[0] for x in ordered]
            centers = [x[1] for x in ordered]

            # Crear límites laterales
            boundaries = [0] + [(centers[i]+centers[i+1])/2 for i in range(len(centers)-1)] + [9999]

            # -----------------------------------------
            # 2) TOMAR SOLO PALABRAS DEBAJO DEL HEADER
            # -----------------------------------------
            header_top = min(w["top"] for w in words)
            data_words = [w for w in words if w["top"] > header_top + 8]
            data_words.sort(key=lambda w: (w["top"], w["x0"]))

            # -----------------------------------------
            # 3) AGRUPAR POR FILAS
            # -----------------------------------------
            rows = []
            current = []
            last_top = None

            for w in data_words:
                if last_top is None or abs(w["top"] - last_top) <= row_tolerance:
                    current.append(w)
                    last_top = w["top"]
                else:
                    rows.append(current)
                    current = [w]
                    last_top = w["top"]

            if current:
                rows.append(current)

            # -----------------------------------------
            # 4) ASIGNAR PALABRAS A COLUMNAS
            # -----------------------------------------
            for r in rows:
                row_dict = {c: "" for c in colnames}
                for w in r:
                    x = w["x0"]
                    col_idx = next(i for i in range(len(boundaries)-1)
                                   if boundaries[i] <= x < boundaries[i+1])
                    col = colnames[col_idx]
                    row_dict[col] += " " + w["text"]

                # cleanup
                for k in row_dict:
                    row_dict[k] = row_dict[k].strip()

                collected_rows.append(row_dict)

    df = pd.DataFrame(collected_rows)

    # -----------------------------------------
    # 5) UNIR MULTILÍNEAS (EURO)
    # -----------------------------------------
    if merge_multiline and multiline_key:
        final_rows = []
        buffer = None

        for _, row in df.iterrows():
            keyval = row[multiline_key].strip()

            if keyval:
                if buffer:
                    final_rows.append(buffer)
                buffer = row.copy()
            else:
                if buffer is None:
                    buffer = row.copy()
                else:
                    for col in headers:
                        buffer[col] = (str(buffer[col]) + " " + str(row[col])).strip()

        if buffer:
            final_rows.append(buffer)

        df = pd.DataFrame(final_rows)

    df = df.applymap(lambda x: x.strip() if isinstance(x, str) else x)

    return df
```

```python
def extractor_euro(pdf_path):
    import re
    import pdfplumber
    import statistics
    import pandas as pd

    results = []

    def prefer_doc(doccol, co, detalle):
        full_txt = ' '.join(filter(None, [doccol, co, detalle])).replace('\n', ' ')
        prefix_match = re.search(r'\b(PMP|DEC|PM|DE|DEV)\b', full_txt, flags=re.IGNORECASE)
        num_match    = re.search(r'\b(\d{3,10})\b', full_txt.replace(',', ''))
        if prefix_match and num_match:
            return f"{prefix_match.group(1).upper()} {num_match.group(1)}"
        if num_match:
            return num_match.group(1)
        return ''

    def clean_co(co_raw):
        if not co_raw: return ''
        toks = str(co_raw).split()
        for t in toks:
            if re.match(r'^[A-Za-z]{2,5}$', t):
                return t
        return toks[0] if toks else ''

    def clean_detalle(co_raw, doccol, detalle_raw, doc_found):
        comb = str(detalle_raw or '').replace('\n', ' ').strip()
        if doc_found:
            comb = re.sub(r'\b' + re.escape(doc_found) + r'\b', ' ', comb)
        comb = re.sub(r'\b\d{3,10}\b', ' ', comb)
        comb = re.sub(r'\s+', ' ', comb).strip()
        return comb

    def extract_num_columns(row, num_cols=['Descuentos','Retenciones','Valor Factura','Valor Pago']):
        text = ' '.join([str(row.get(col,'')).replace('\n',' ').replace('\r',' ').strip() for col in num_cols])
        nums = re.findall(r'\$?-?\d[\d,\.]*', text)
        nums = [n.replace('$','').replace(' ','') for n in nums]
        result = {}
        for i, col in enumerate(num_cols):
            result[col] = nums[i] if i < len(nums) else ''
        return result

    def merge_doc_prefix_rows(parsed, doc_field='Doc.Cruce'):
        merged = []
        i = 0
        while i < len(parsed):
            row = parsed[i].copy()
            doc = str(row.get(doc_field, '')).strip()
            if re.match(r'^[A-Za-z]{2,4}$', doc, flags=re.IGNORECASE) and i + 1 < len(parsed):
                nxt = parsed[i+1]
                nxt_doc = str(nxt.get(doc_field, '')).strip()
                if re.match(r'^\d{3,10}$', nxt_doc):
                    for k in list(row.keys()):
                        if k == 'page':
                            continue
                        row[k] = (str(row.get(k, '')).strip() + ' ' + str(nxt.get(k, '')).strip()).strip()
                    merged.append(row)
                    i += 2
                    continue
            merged.append(row)
            i += 1
        return merged

    with pdfplumber.open(pdf_path) as pdf:
        for pnum, page in enumerate(pdf.pages, start=1):
            words = page.extract_words()
            if not words:
                continue

            header_cands = []
            for w in words:
                t = re.sub(r'\s+', ' ', w['text']).strip()
                tl = t.lower().replace(' ', '')
                if tl in ('reg','reg.') or 'detalle' in tl or 'doc' in tl or 'cruce' in tl or tl in ('descuentos','retenciones','valorfactura','valorpago'):
                    header_can_
```
