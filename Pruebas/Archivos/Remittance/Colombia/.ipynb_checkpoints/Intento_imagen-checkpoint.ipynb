```python
import pytesseract
from PIL import Image
import pandas as pd
import re

# --------------------------------------------------
# 1) Ruta del archivo
# --------------------------------------------------

image_path = "Remittance_Cruz_Verde.png"
img = Image.open(image_path)

# --------------------------------------------------
# 2) OCR
# --------------------------------------------------

raw_text = pytesseract.image_to_string(img, lang="spa")

lines = raw_text.split("\n")

facturas = []
importes = []

# Regex para números de factura
regex_fact = r"\b(PMP\d+|NCMI\d+)\b"

# Regex para importes válidos
regex_importe = r"\(?\d{1,3}(?:,\d{3})+\)?"

for line in lines:
    mf = re.search(regex_fact, line)
    mi = re.search(regex_importe, line)

    if mf and mi:
        facturas.append(mf.group(0))
        importes.append(mi.group(0))

# --------------------------------------------------
# 3) Limpieza del importe
# --------------------------------------------------

def limpiar_importe(x):
    x = x.replace(",", "")
    if x.startswith("(") and x.endswith(")"):
        return -int(x.strip("()"))
    return int(x)

importes_limpios = [limpiar_importe(i) for i in importes]

# --------------------------------------------------
# 4) DataFrame final
# --------------------------------------------------

df = pd.DataFrame({
    "Referencia / Factura": facturas,
    "Importe de Factura": importes_limpios
})

df
```
