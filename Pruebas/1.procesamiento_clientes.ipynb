```python
import os
import ipywidgets as widgets
from IPython.display import display

# ==============================================
# Importar clientes de Colombia y Ecuador
# ==============================================
from clientes.Colombia import (
    olimpica,
    farmatodo,
    D1,
    Cencosud,
    copi,
    Euro,
    oxxo
)

from clientes.Ecuador import (
    El_Rosado,
    Tia
)

# ==============================================
# Registro de clientes por país
# ==============================================
CLIENTES = {
    "Colombia": {
        "Olimpica": olimpica.procesar,
        "Farmatodo": farmatodo.procesar,
        "D1": D1.procesar,
        "Cencosud": Cencosud.procesar,
        "Copidrogas": copi.procesar,
        "Euro": Euro.procesar,
        "Oxxo": oxxo.procesar
    },
    "Ecuador": {
        "El Rosado": El_Rosado.procesar,
        "Tia": Tia.procesar
    }
}

# ==============================================
# Widgets interactivos
# ==============================================
dropdown_pais = widgets.Dropdown(
    options=list(CLIENTES.keys()),
    description="País:"
)

dropdown_cliente = widgets.Dropdown(
    description="Cliente:"
)

button = widgets.Button(description="Procesar")
output = widgets.Output()

# ==============================================
# Actualizar clientes según país
# ==============================================
def actualizar_clientes(change):
    pais = change["new"]
    clientes_disponibles = list(CLIENTES[pais].keys())
    dropdown_cliente.options = clientes_disponibles
    if clientes_disponibles:
        dropdown_cliente.value = clientes_disponibles[0]

dropdown_pais.observe(actualizar_clientes, names="value")
actualizar_clientes({"new": dropdown_pais.value})

display(dropdown_pais, dropdown_cliente, button, output)

# ==============================================
# Función principal al presionar el botón
# ==============================================
def on_click(b):
    with output:
        output.clear_output()
        try:
            pais = dropdown_pais.value
            cliente = dropdown_cliente.value
            procesar = CLIENTES[pais][cliente]
            result = procesar()

            print(f"✅ Proceso completado para {cliente} ({pais})")

            # Si el resultado es un diccionario de DataFrames
            if isinstance(result, dict):
                for key, df in result.items():
                    print(f"\n📄 {key}:")
                    display(df.head())
            # Si el resultado es una lista o tupla de DataFrames
            elif isinstance(result, (list, tuple)):
                for i, df in enumerate(result, start=1):
                    print(f"\n📄 DataFrame {i}:")
                    display(df.head())
            # Si es un solo DataFrame
            else:
                display(result.head())

        except Exception as e:
            import traceback; traceback.print_exc()

button.on_click(on_click)
```
