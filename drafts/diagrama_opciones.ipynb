```python
from PIL import Image, ImageDraw, ImageFont
import textwrap

# --- Configuración de imagen ---
W, H = 1000, 500
bg_color = (245, 245, 245)
img = Image.new("RGB", (W, H), bg_color)
draw = ImageDraw.Draw(img)

# --- Fuente más grande ---
try:
    font_title = ImageFont.truetype("DejaVuSans-Bold.ttf", 28)
    font_option = ImageFont.truetype("DejaVuSans-Bold.ttf", 18)
    font_desc = ImageFont.truetype("DejaVuSans.ttf", 17)
except:
    font_title = ImageFont.load_default()
    font_option = ImageFont.load_default()
    font_desc = ImageFont.load_default()

# --- Título ---
title = "Opciones de carga de archivos en la aplicación"
w, h = draw.textsize(title, font=font_title)
draw.text(((W-w)/2, 20), title, font=font_title, fill=(0,0,0))

# --- Columnas / Opciones ---
options = [
    ("Arrastrar y soltar", "Arrastrar el archivo o subirle mediante un click", (135,206,250)),
    ("Carpeta local", "Colocar archivos en la carpeta local predefinida", (144,238,144)),
    ("SharePoint", "Subir archivos a un SharePoint gestionado por Unilever", (255,182,193))
]

col_width = W // 3
box_w, box_h = 280, 140  # un poco más grandes para las fuentes
start_y = 100

for i, (title_opt, desc, color) in enumerate(options):
    x_center = i * col_width + col_width//2
    x0, y0 = x_center - box_w//2, start_y
    x1, y1 = x0 + box_w, y0 + box_h

    # Fondo de la columna
    draw.rounded_rectangle((x0, y0, x1, y1), radius=15, fill=color, outline=(100,100,100), width=2)

    # Título de la opción
    w_opt, h_opt = draw.textsize(title_opt, font=font_option)
    draw.text((x_center - w_opt/2, y0 + 20), title_opt, font=font_option, fill=(0,0,0))

    # Descripción centrada, evitando cortar palabras
    wrapper = textwrap.TextWrapper(width=35, break_long_words=False)  # ancho ajustado
    lines = wrapper.wrap(desc)
    total_h = len(lines) * 22
    for j, line in enumerate(lines):
        w_line, h_line = draw.textsize(line, font=font_desc)
        draw.text((x_center - w_line/2, y0 + 60 + j*22), line, font=font_desc, fill=(50,50,50))

# --- Rectángulo final ---
rect_w, rect_h = 450, 90
rect_x0, rect_y0 = W//2 - rect_w//2, 380
rect_x1, rect_y1 = rect_x0 + rect_w, rect_y0 + rect_h
draw.rounded_rectangle((rect_x0, rect_y0, rect_x1, rect_y1), radius=15, fill=(0,76,151))

final_text = "Generación automática de templates HRC y envío a Outlook"
w_final, h_final = draw.textsize(final_text, font=font_option)
draw.text((W//2 - w_final/2, rect_y0 + (rect_h-h_final)/2), final_text, font=font_option, fill=(255,255,255))

# Flechas desde cada columna
for i in range(3):
    x_start = i * col_width + col_width//2
    draw.line((x_start, start_y + box_h, W//2, rect_y0), fill=(0,0,0), width=3)
    draw.polygon([(W//2-5, rect_y0-5), (W//2+5, rect_y0-5), (W//2, rect_y0)], fill=(0,0,0))

# --- Guardar imagen ---
output_path = "imagenes/diagrama_opciones.png"
img.save(output_path, format="PNG")
print(f"✅ Diagrama actualizado con fuentes más grandes: {output_path}")
```
