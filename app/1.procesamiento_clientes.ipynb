```python
import ipywidgets as widgets
from IPython.display import display
from clientes import olimpica, farmatodo, D1

# Registro de clientes
CLIENTES = {
    "Olimpica": olimpica.procesar,
    "Farmatodo": farmatodo.procesar,
    "D1": D1.procesar
}

# Widgets
dropdown = widgets.Dropdown(options=list(CLIENTES.keys()), description="Cliente:")
button = widgets.Button(description="Procesar")
output = widgets.Output()

display(dropdown, button, output)

def on_click(b):
    with output:
        output.clear_output()
        try:
            cliente = dropdown.value
            procesar = CLIENTES[cliente]
            df = procesar()
            print(f"✅ Proceso completado para {cliente}")
            display(df.head())
        except Exception as e:
            import traceback; traceback.print_exc()

button.on_click(on_click)
```
