```python
from PIL import Image, ImageDraw, ImageFont

# --- Configuración base ---
W, H = 500, 400
bg_color = (30, 30, 30)  # fondo oscuro
img = Image.new("RGB", (W, H), bg_color)
draw = ImageDraw.Draw(img)

# --- Logo real ---
try:
    logo = Image.open("Archivos/logo_unilever.png").convert("RGBA")
    logo = logo.resize((120, 120))
    img.paste(logo, ((W-120)//2, 15), logo)  # centrado arriba con transparencia
except Exception as e:
    print("⚠️ No se pudo cargar el logo:", e)

# --- Fuente ---
try:
    font_title = ImageFont.truetype("DejaVuSans.ttf", 15)
except Exception:
    font_title = ImageFont.load_default()

# --- Botón Procesamiento masivo (arriba del combo) ---
btn_w, btn_h = 300, 50
btn_x0 = 100
btn_y0 = 150  # arriba del combo
for i in range(btn_h):
    color = (0, 76+i, min(151+i, 255))
    draw.line((btn_x0, btn_y0+i, btn_x0+btn_w, btn_y0+i), fill=color)
draw.rounded_rectangle((btn_x0, btn_y0, btn_x0+btn_w, btn_y0+btn_h),
                       radius=10, outline=(0,0,0), width=1)
try:
    font_btn = ImageFont.truetype("DejaVuSans-Bold.ttf", 16)
except Exception:
    font_btn = ImageFont.load_default()
btn_text = "Procesamiento masivo"
w_btn, h_btn = draw.textsize(btn_text, font=font_btn)
draw.text((btn_x0 + (btn_w-w_btn)/2, btn_y0 + (btn_h-h_btn)/2 - 2),
          btn_text, font=font_btn, fill="white")

# --- Texto descriptivo ---
title = "Seleccione el cliente para generar el Template HRC"
w, h = draw.textsize(title, font=font_title)
draw.text(((W-w)/2, btn_y0 + btn_h + 15), title, font=font_title, fill=(230,230,230))

# --- Combobox ---
cb_x0, cb_y0 = 75, btn_y0 + btn_h + 45
cb_w, cb_h = 350, 40
draw.rounded_rectangle((cb_x0, cb_y0, cb_x0+cb_w, cb_y0+cb_h),
                       radius=6, outline=(100,100,100), width=2, fill=(50,50,50))
try:
    font_cb = ImageFont.truetype("DejaVuSans.ttf", 13)
except Exception:
    font_cb = ImageFont.load_default()
draw.text((cb_x0+12, cb_y0+11), "Seleccionar cliente...", font=font_cb, fill=(180,180,180))
# Flecha del combo
arrow_x = cb_x0 + cb_w - 24
draw.polygon([(arrow_x, cb_y0+16), (arrow_x+10, cb_y0+16), (arrow_x+5, cb_y0+24)], fill=(180,180,180))

# --- Botón Procesar individualmente (debajo del combo) ---
btn_y0_ind = cb_y0 + cb_h + 15
for i in range(btn_h):
    color = (0, 76+i, min(151+i, 255))
    draw.line((btn_x0, btn_y0_ind+i, btn_x0+btn_w, btn_y0_ind+i), fill=color)
draw.rounded_rectangle((btn_x0, btn_y0_ind, btn_x0+btn_w, btn_y0_ind+btn_h),
                       radius=10, outline=(0,0,0), width=1)
btn_text_ind = "Procesar individualmente"
w_btn, h_btn = draw.textsize(btn_text_ind, font=font_btn)
draw.text((btn_x0 + (btn_w-w_btn)/2, btn_y0_ind + (btn_h-h_btn)/2 - 2),
          btn_text_ind, font=font_btn, fill="white")

# --- Footer ---
footer_text = "Unilever - Automatización HRC"
try:
    font_footer = ImageFont.truetype("DejaVuSans.ttf", 11)
except Exception:
    font_footer = ImageFont.load_default()
fw, fh = draw.textsize(footer_text, font=font_footer)
draw.text(((W-fw)/2, H-22), footer_text, font=font_footer, fill=(160,160,160))

# --- Guardar archivo ---
output_path = "Bosquejo_3.png"
img.save(output_path, format="PNG")
print(f"✅ Mockup actualizado con botón masivo arriba: {output_path}")
```
