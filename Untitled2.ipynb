```python
import tkinter as tk
from tkinter import messagebox
from datetime import datetime
import clientes.farmatodo as farmatodo
import clientes.olimpica as olimpica
from PIL import Image, ImageTk  # pip install pillow

def procesar_cliente(cliente):
    if cliente == "Farmatodo":
        return farmatodo.procesar()
    elif cliente == "Olimpica":
        return olimpica.procesar()
    else:
        raise ValueError("Cliente no soportado")

def ejecutar(cliente):
    try:
        resultado = procesar_cliente(cliente)

        # Fecha para el nombre del archivo
        fecha = datetime.today().strftime("%Y%m%d")
        archivo_salida = f"Template_HRC_{cliente}_{fecha}.xlsx"

        # Exportar directamente desde la función de cliente
        resultado.to_excel(archivo_salida, index=False)

        messagebox.showinfo(
            "Proceso finalizado ✅",
            f"El template para {cliente} fue generado correctamente.\n\n"
            f"Archivo exportado: {archivo_salida}"
        )
    except Exception as e:
        messagebox.showerror("Error ❌", f"Ocurrió un problema:\n{e}")

# --- Ventana principal ---
root = tk.Tk()
root.title("Generador de Template HRC")
root.geometry("500x400")

# --- Logo Unilever ---
try:
    logo = Image.open("logo_unilever.png")  # guardá el logo en la misma carpeta
    logo = logo.resize((120, 120))
    logo_tk = ImageTk.PhotoImage(logo)
    lbl_logo = tk.Label(root, image=logo_tk)
    lbl_logo.pack(pady=10)
except Exception as e:
    print("No se pudo cargar el logo:", e)

# Texto descriptivo
label = tk.Label(
    root,
    text="Seleccione el cliente para generar el Template HRC",
    font=("Arial", 12),
    pady=10
)
label.pack()

# Botones de clientes
btn_farmatodo = tk.Button(
    root, text="Farmatodo", font=("Arial", 14), width=25, bg="#004C97", fg="white",
    command=lambda: ejecutar("Farmatodo")
)
btn_farmatodo.pack(pady=10)

btn_olimpica = tk.Button(
    root, text="Olimpica", font=("Arial", 14), width=25, bg="#004C97", fg="white",
    command=lambda: ejecutar("Olimpica")
)
btn_olimpica.pack(pady=10)

# Footer
footer = tk.Label(root, text="Unilever - Automatización HRC", font=("Arial", 9), fg="gray")
footer.pack(side="bottom", pady=5)

# Ejecutar ventana
root.mainloop()
```
