```python
import os
from datetime import datetime
import clientes.farmatodo as farmatodo
import clientes.olimpica as olimpica

def procesar_cliente(cliente):
    if cliente == "Farmatodo":
        return farmatodo.procesar()
    elif cliente == "Olimpica":
        return olimpica.procesar()
    else:
        raise ValueError("Cliente no soportado")

def main():
    print("=====================================")
    print("     🚀 Generador de Template HRC     ")
    print("=====================================")
    print("Seleccione un cliente:")
    print("1 - Farmatodo")
    print("2 - Olimpica")
    print("-------------------------------------")

    opcion = input("Ingrese opción (1 o 2): ")

    if opcion == "1":
        cliente = "Farmatodo"
    elif opcion == "2":
        cliente = "Olimpica"
    else:
        print("❌ Opción inválida")
        return

    print(f"\n🔄 Procesando datos para {cliente}...")
    resultado = procesar_cliente(cliente)

    fecha = datetime.today().strftime("%Y%m%d")
    archivo_salida = f"Template_HRC_{cliente}_{fecha}.xlsx"

    resultado.to_excel(archivo_salida, index=False)
    print(f"✅ Proceso finalizado. Archivo exportado: {archivo_salida}")

if __name__ == "__main__":
    main()
```
