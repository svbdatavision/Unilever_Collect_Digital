```python
from PIL import Image, ImageDraw, ImageFont
import textwrap

# --- Configuración base ---
W, H = 500, 500
bg_color = (30, 30, 30)
img = Image.new("RGB", (W, H), bg_color)
draw = ImageDraw.Draw(img)

# --- Logo real ---
try:
    logo = Image.open("Archivos/logo_unilever.png").convert("RGBA")
    logo = logo.resize((120, 120))
    img.paste(logo, ((W-120)//2, 15), logo)
except Exception as e:
    print("⚠️ No se pudo cargar el logo:", e)

# --- Texto descriptivo ---
try:
    font_title = ImageFont.truetype("DejaVuSans.ttf", 15)
except Exception:
    font_title = ImageFont.load_default()
title = "Seleccione el cliente para generar el Template HRC"
w, h = draw.textsize(title, font=font_title)
draw.text(((W-w)/2, 150), title, font=font_title, fill=(230,230,230))

# --- Combobox ---
cb_x0, cb_y0 = 75, 190
cb_w, cb_h = 350, 40
draw.rounded_rectangle((cb_x0, cb_y0, cb_x0+cb_w, cb_y0+cb_h),
                       radius=6, outline=(100,100,100), width=2, fill=(50,50,50))
try:
    font_cb = ImageFont.truetype("DejaVuSans.ttf", 13)
except Exception:
    font_cb = ImageFont.load_default()
draw.text((cb_x0+12, cb_y0+11), "Seleccionar cliente...", font=font_cb, fill=(180,180,180))
arrow_x = cb_x0 + cb_w - 24
draw.polygon([(arrow_x, cb_y0+16), (arrow_x+10, cb_y0+16), (arrow_x+5, cb_y0+24)], fill=(180,180,180))

# --- Cajas de archivos ---
box_w, box_h = 350, 60
start_y = 250
spacing = 20
file_boxes = [("Remittance", start_y), ("FBL5N", start_y + box_h + spacing)]
try:
    font_box = ImageFont.truetype("DejaVuSans.ttf", 12)  # tamaño un poco más pequeño
except:
    font_box = ImageFont.load_default()

for label, y in file_boxes:
    draw.rounded_rectangle((cb_x0, y, cb_x0+box_w, y+box_h),
                           radius=6, outline=(120,120,120), width=2, fill=(50,50,50))
    # Texto centrado y ajustado al ancho del box
    text = f"Arrastre aquí el archivo de {label} o haga clic para adjuntar el Archivo"
    # Ajuste de texto a varias líneas si es necesario
    max_chars_per_line = 40
    lines = textwrap.wrap(text, width=max_chars_per_line)
    total_text_height = len(lines) * font_box.getsize('A')[1]
    for i, line in enumerate(lines):
        tw, th = draw.textsize(line, font=font_box)
        draw.text((cb_x0 + (box_w-tw)/2, y + (box_h - total_text_height)/2 + i*th),
                  line, font=font_box, fill=(200,200,200))

# --- Botón Generar ---
btn_x0, btn_y0 = 100, start_y + 2*(box_h + spacing) + 10
btn_w, btn_h = 300, 52
for i in range(btn_h):
    color = (0, 76+i, min(151+i, 255))
    draw.line((btn_x0, btn_y0+i, btn_x0+btn_w, btn_y0+i), fill=color)
draw.rounded_rectangle((btn_x0, btn_y0, btn_x0+btn_w, btn_y0+btn_h),
                       radius=10, outline=(0,0,0), width=1)
try:
    font_btn = ImageFont.truetype("DejaVuSans-Bold.ttf", 18)
except Exception:
    font_btn = ImageFont.load_default()
btn_text = "Generar Template"
w_btn, h_btn = draw.textsize(btn_text, font=font_btn)
draw.text((btn_x0 + (btn_w-w_btn)/2, btn_y0 + (btn_h-h_btn)/2 - 2),
          btn_text, font=font_btn, fill="white")

# --- Footer ---
footer_text = "Unilever - Automatización HRC"
try:
    font_footer = ImageFont.truetype("DejaVuSans.ttf", 11)
except Exception:
    font_footer = ImageFont.load_default()
fw, fh = draw.textsize(footer_text, font=font_footer)
draw.text(((W-fw)/2, H-22), footer_text, font=font_footer, fill=(160,160,160))

# --- Guardar archivo ---
output_path = "Bosquejo_2.png"
img.save(output_path, format="PNG")
print(f"✅ Mockup generado con texto centrado en boxes: {output_path}")
```
