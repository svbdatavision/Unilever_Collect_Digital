```python
import ipywidgets as widgets
from IPython.display import display

from clientes import olimpica, farmatodo  # importamos reglas externas
```

```python
# Defino lista de clientes
clientes = ["Farmatodo", "Olimpica"]
```

```python
# Dropdown
clientes = ["Farmatodo", "Olimpica"]
dropdown_cliente = widgets.Dropdown(
    options=clientes,
    description="Cliente:"
)
display(dropdown_cliente)
```

```python
# Función para ejecutar
def procesar_cliente(cliente):
    if cliente == "Farmatodo":
        return farmatodo.procesar()
    elif cliente == "Olimpica":
        return olimpica.procesar()
    else:
        raise ValueError("Cliente no soportado")
```

```python
# Ejecutar
resultado = procesar_cliente(dropdown_cliente.value)
#resultado
```
